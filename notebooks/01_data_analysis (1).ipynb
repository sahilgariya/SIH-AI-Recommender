```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("/content/sih_2026_problem_statements.csv")
```

```python
df.info()
```

```python
# CHECK NULL-VALUE
df.isnull().sum()
```

```python
#  CHECK THE DUPLICATE VALUE
df.duplicated().sum()
```

```python
# CLEAN EXTRA-SPACE FROM TEXT COLUMN
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
# REMOVING THE UNNECESSARY INDEX COLUMN
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# CONVERT DEADLINE-COLUMN TO DATE (CHANGE OBJECT TO DATE)
df['Deadline for Idea Submission'] = pd.to_datetime(
    df['Deadline for Idea Submission'],
    errors='coerce'
)
```

```python
# SAVE DATASET
df.to_csv("cleaned_sih_2026_problem_statements.csv", index=False)
```

```python
# COMBINE 'TITLE' & 'THEME' & 'CATEGORY' & 'DETAILS'
df["combined_text"] = (
    df["Problem Statement Title"].fillna("").astype(str) + " " +
    df["Theme"].fillna("").astype(str) + " " +
    df["Category"].fillna("").astype(str) + " " +
    df["Details"].fillna("").astype(str)
)

# REMOVE EXTRA SPACES
df["combined_text"] = df["combined_text"].str.strip()

# SAVE THE UPDATED DATASET
df.to_csv("cleaned_dataset.csv", index=False)


```

```python
# HANDLE MISSING VALUES
df["combined_text"] = df["combined_text"].fillna("")
```

```python
# CREATE  THE TF-IDF VECTORIZER
tfidf = TfidfVectorizer(
    stop_words="english",
    lowercase=True
)

# CONVERT 'COMBINED_TEXT' INTO NUMERICALS VECTORS
tfidf_matrix = tfidf.fit_transform(df["combined_text"])

# CHECK MATRIX SHAPE
print("Dataset TF-IDF shape:", tfidf_matrix.shape)

# USER ENTERS SKILLS/INTERSESTS
user_skills = input("Enter your skills and interests: ")
# "Python Machine Learning Computer Vision Agriculture"
print("User entered:", user_skills)


# CONVERTER USER SKILLS INTO A VECTORS
user_vector = tfidf.transform([user_skills])

# CHECKS USER VECTOR SHAPE
print("User vector shape:", user_vector.shape)
```

```python
# CALCULATED THE SIMILARITY SCORES
similarity_scores = cosine_similarity(user_vector, tfidf_matrix)
```

```python
# CHECK THE SHAPE OF THE SIMILARITY RESULT
similarity_scores.shape
```

```python
# CONVERT THE RESULT
similarity_scores = similarity_scores.flatten()
```

```python
# TOP 10 INDICES
top_10_indices = np.argsort(similarity_scores)[-10:][::-1]
```

```python
# DISPLAY TOP 10 SIMILARITY SCORE
# TOP 10 SIH PROBLEM RECOMMENDATIONS

print("=" * 100)
print("TOP 10 SIH PROBLEM RECOMMENDATIONS")
print("=" * 100)

for rank, index in enumerate(top_10_indices, start=1):

    # Get complete problem data
    problem = df.iloc[index]

    # Get real cosine similarity score
    score = similarity_scores[index]

    # Convert score to percentage
    score_percentage = score * 100

    print(f"\n{'=' * 100}")
    print(f"RANK {rank}")
    print(f"{'=' * 100}")

    print("\nProblem Title:")
    print(problem["Problem Statement Title"])

    print("\nPS Number:")
    print(problem["PS Number"])

    print("\nTheme:")
    print(problem["Theme"])

    print("\nCategory:")
    print(problem["Category"])

    print("\nFull Problem Description:")
    print(problem["Details"])

    print(f"\nAI Match Score: {score_percentage:.2f}%")

    print("\n" + "-" * 100)
```
